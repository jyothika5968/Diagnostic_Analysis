```python
# E commerce Sales cart
# Diagnostic Analysis
import mat
customer_name=["Jyothi","Chinnu","Naani","Deepak"]
internet_speed=[2,5,8,6]
Payment_failure=[0,3,1,4,]
Delivery_Charges=[20,40,60,80]
purchase_completed=[True,False,True,True]
# Display data set
print("----- E-Commerce Sales Cart-----")
for i in range(len(customer_name)):
    print("\nCustomer Name:",customer_name[i],"Internet Speed:",internet_speed[i],"Payment Failure:",Payment_failure[i],"Delivery Charges:",Delivery_Charges[i])
 # to Find total abandment
```

```python
# to find total abondonment
total_abandonment = purchase_completed.count(True)
print(f"Total Abandonment: {total_abandonment}")
```

```python
# METHOD 1
# to find internet diagnostics
print("\n Customer with slow internet")
for i in range(len(customer_name)):
  if internet_speed[i] < 3:
    print(f"{customer_name[i],"-",internet_speed[i],"Mbps"}")
```

```python
# to find payment diagnostics
print("\n customer with payment failure")
for i in range(len(customer_name)):
  if Payment_failure[i] > 0:
    print(f"{customer_name[i]} - {Payment_failure[i]}")
```

```python
# to find delivery diagnostics
print("\n customer with high delivery charges")
for i in range(len(customer_name)):
  if Delivery_Charges[i] > 50:
    print(f"{customer_name[i]} - {Delivery_Charges[i]}")
```

```python
# METHOD 2
#internet diagnosis
print('\n--- Internet Speed Diagnosis ---')
for i in range(len(customer_name)):
  customer = customer_name[i]
  internet_spd = internet_speed[i]
  purchased = purchase_completed[i]

  print(f"\nCustomer: {customer}")
  print(f"  Internet Speed: {internet_spd}")
  print(f"  Purchased Completed: {purchased}")

  if purchased == 'no' and internet_spd < 3:
    print("  Diagnosis: Customer abandoned purchase, and slow internet might be a contributing factor.")
  elif purchased == 'no' and internet_spd >= 3:
    print("  Diagnosis: Customer abandoned purchase, but internet speed was not a reported issue.")
  else:
    print("  Diagnosis: Purchase completed. Internet speed seems adequate.")
```

```python
# METHOD 2
print('\n--- Payment Diagnosis ---')
for i in range(len(customer_name)):
  customer = customer_name[i]
  current_payment_fail = Payment_failure[i] # Corrected variable name
  purchased = purchase_completed[i] # Corrected variable name

  print(f"\nCustomer: {customer}")
  print(f"  Payment Failure Count: {current_payment_fail}")
  print(f"  Purchased Completed: {purchased}")

  if not purchased and current_payment_fail > 0: # Corrected comparison
    print("  Diagnosis: Customer abandoned purchase, and payment failures might be the reason.")
  elif not purchased and current_payment_fail == 0: # Corrected comparison
    print("  Diagnosis: Customer abandoned purchase, but payment was not a reported issue.")
  else:
    print("  Diagnosis: Purchase completed. Payment was successful.")
```

```python
# METHOD 2
print('\n--- Delivery Diagnosis ---')
for i in range(len(customer_name)):
  customer = customer_name[i]
  current_delivery_charge = Delivery_Charges[i] # Corrected variable name
  purchased = purchase_completed[i] # Corrected variable name

  print(f"\nCustomer: {customer}")
  print(f"  Delivery Charges: {current_delivery_charge}")
  print(f"  Purchased Completed: {purchased}")

  if not purchased and current_delivery_charge > 30: # Assuming charges > 30 are high enough to deter and corrected comparison
    print("  Diagnosis: Customer abandoned purchase, and high delivery charges might be a contributing factor.")
  elif not purchased and current_delivery_charge <= 30: # Corrected comparison
    print("  Diagnosis: Customer abandoned purchase, but delivery charges were not excessively high.")
  else:
    print("  Diagnosis: Purchase completed. Delivery charges seem acceptable.")
```

```python
# ROOT CAUSE ANALYSIS
print("\n Root Cause Analysis")
for i in range (len(customer_name)):
  if(purchase_completed[i])== False:
    if internet_speed[i] < 5 :
      print(customer_name[i],"abandoned cart due to slow internet")
    elif Payment_failure[i]>0:
      print(customer_name[i],"abandoned cart due to payment failure")
    elif Delivery_Charges[i]>50:
      print(customer_name[i],"abandoned cart due to high delivery charges")
```

```python
# Visualization
import matplotlib.pyplot as plt
plt.bar(customer_name,internet_speed)
plt.xlabel("Customer Name")
plt.ylabel("Internet Speed")
plt.title("Internet Speed Analysis")
plt.show()
```
