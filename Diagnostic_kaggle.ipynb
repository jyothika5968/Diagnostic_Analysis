```python
import pandas as pd
df=pd.read_csv("/content/customer_churn_dataset-training-master.csv",encoding='latin -1')
df.tail()
```

```python
# finding abandnment
abandnment = df['Churn'].sum()
print("---total abandnment---")
print(f"Total abandonment: {abandnment}")
```

```python
#network diagnostics
for i in range(len(df['CustomerID'])):
  if df['Total Spend'][i] < 500 :
     print(df['CustomerID'][i],"-",df['Total Spend'][i] )
```

```python
## to find payment diagnostics
print("\n customer with payment Delay")
for i in range(len(df['CustomerID'])):
  if df['Payment Delay'] [i] > 0:
    print(df['CustomerID'][i] - df['Payment Delay'][i])
```

```python
# to find delivery Diagnostics
print("\n customer with high delivery charges")
for i in range(len(df['CustomerID'])):
  if df['Last Interaction'] [i] > 20:
    print(df['Last Interaction'][i],"-",df['CustomerID'][i])
```

```python
# ROOT CAUSE ANALYSIS
print("\n Root Cause Analysis")
for i in range (len(df['CustomerID'])):
  if(df['Churn'][i])== 1:
    if df['Total Spend'][i] < 500 :
      print(df['CustomerID'][i],"abandoned cart due to low spend")
    elif df['Payment Delay'][i]>0:
      print(df['CustomerID'][i],"abandoned cart due to payment Delay")
    elif df['Last Interaction'] [i]>20:
      print(df['CustomerID'] [i],"abandoned cart due to over Interaction")
```

```python
# Visualization
import matplotlib.pyplot as plt
plt.bar(df['CustomerID'],df['Total Spend'])
plt.xlabel(df['CustomerID'])
plt.ylabel(df['Total Spend'])
plt.title("Internet Speed Analysis")
plt.show()
```
